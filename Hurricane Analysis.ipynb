3. In addition to organizing the hurricanes in a dictionary with names as the key, you want to be able to organize the hurricanes by year.

   Write a function that converts the current dictionary of hurricanes to a new dictionary, where the keys are years and the values are lists containing a dictionary for each hurricane that occurred in that year.
   
   For example, the key `1932` would yield the value: `[{'Name': 'Bahamas', 'Month': 'September', 'Year': 1932, 'Max Sustained Wind': 160, 'Areas Affected': ['The Bahamas', 'Northeastern United States'], 'Damage': 'Damage not recorded', 'Deaths': 16}, {'Name': 'Cuba II', 'Month': 'November', 'Year': 1932, 'Max Sustained Wind': 175, 'Areas Affected': ['Lesser Antilles', 'Jamaica', 'Cayman Islands', 'Cuba', 'The Bahamas', 'Bermuda'], 'Damage': 40000000.0, 'Deaths': 3103}]`.
   
   Test your function on your hurricane dictionary.

````python
# 3
# Organizing by Year

def organize_by_year(hurricanes):
    year_dict = {}
    for hurricane in hurricanes.values():
        year = hurricane["Year"]

        if year not in year_dict:
            year_dict[year] = []

        year_dict[year].append(hurricane)

    return year_dict

# create a new dictionary of hurricanes with year and key
hurricanes_by_year = organize_by_year(hurricane_dict)
print(hurricanes_by_year)
````

4. You believe that knowing how often each of the areas of the Atlantic are affected by these strong hurricanes is important for making preparations for future hurricanes.

   Write a function that counts how often each area is listed as an affected area of a hurricane. Store and return the results in a dictionary where the keys are the affected areas and the values are counts of how many times the areas were affected.
   
   Test your function on your hurricane dictionary.

````python
# 4
# Counting Damaged Areas
def count_damaged_areas(hurricanes):
    damaged_areas_count = {}
    for hurricane in hurricanes.values():
        for area in hurricane["Arreas affected"]:
            if area in damaged_areas_count:
                damaged_areas_count[area] += 1
            else:
                damaged_areas_count[area] = 1
    return damaged_areas_count 

# create dictionary of areas to store the number of hurricanes involved in
areas_damaged = count_damaged_areas(hurricane_dict)
print(areas_damaged)
````

7. Just as hurricanes are rated by their windspeed, you want to try rating hurricanes based on other metrics.

   Write a function that rates hurricanes on a mortality scale according to the following ratings, where the key is the rating and the value is the upper bound of deaths for that rating.
   
   ```py
   mortality_scale = {0: 0,
   1: 100,
   2: 500,
   3: 1000,
   4: 10000}
   ```
   
   For example, a hurricane with a `1` mortality rating would have resulted in greater than `0` but less than or equal to `100` deaths. A hurricane with a `5` mortality would have resulted in greater than `10000` deaths.
   
   Store the hurricanes in a new dictionary where the keys are the mortaility ratings and the values are lists containing a dictionary for each hurricane that falls into that mortality rating.
   
   Test your function on your hurricane dictionary.

````python
# 7
# Rating Hurricanes by Mortality
mortality_scale = {0: 0,
   1: 100,
   2: 500,
   3: 1000,
   4: 10000}

def hurricane_mortality(hurricanes, mortality_scale):
    hurricanes_mortality = {rating: [] for rating in range(6)}
    for hurricane in hurricanes.values():
        deaths = hurricane["Deaths"]

        if deaths == 0:
            hurricanes_mortality[0].append(hurricane)
            continue

        for rating, limit in mortality_scale.items():
            if rating > 0 and deaths <= limit:
                hurricanes_mortality[rating].append(hurricane)
                break
        else:
            hurricanes_mortality[5].append(hurricane)
    return hurricanes_mortality

# categorize hurricanes in new dictionary with mortality severity as key
print(hurricane_mortality(hurricane_dict, mortality_scale))
````

9. Lastly, you want to rate hurricanes according to how much damage they cause.

   Write a function that rates hurricanes on a damage scale according to the following ratings, where the key is the rating and the value is the upper bound of damage for that rating.
   ```py
   damage_scale = {0: 0,
   1: 100000000,
   2: 1000000000,
   3: 10000000000,
   4: 50000000000}
   ```
   
   For example, a hurricane with a `1` damage rating would have resulted in damages greater than `0` USD but less than or equal to `100000000` USD. A hurricane with a `5` damage rating would have resulted in damages greater than `50000000000` USD (talk about a lot of money).
   
   Store the hurricanes in a new dictionary where the keys are damage ratings and the values are lists containing a dictionary for each hurricane that falls into that damage rating.
   
   Test your function on your hurricane dictionary.

````python
# 9
# Rating Hurricanes by Damage
damage_scale = {0: 0,
                1: 100000000,
                2: 1000000000,
                3: 10000000000,
                4: 50000000000}

# categorize hurricanes in new dictionary with damage severity as key
def damage_rating(hurricanes, damage_scale):
    hurricanes_damages = {rating: [] for rating in range(6)}
    for hurricane in hurricanes.values():
        damages = hurricane["Damages"]

        if damages == "Damages not recorded":
            continue
        
        if damages == 0:
            hurricanes_damages[0].append(hurricane)
            continue

        for rating, limit in damage_scale.items():
            if rating > 0 and damages <= limit:
                hurricanes_damages[rating].append(hurricane)
                break
        else:
            hurricanes_damages[5].append(hurricane)
    return hurricanes_damages

print(damage_rating(hurricane_dict, damage_scale))
````
